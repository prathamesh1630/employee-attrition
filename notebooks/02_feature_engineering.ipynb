```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# Load data
df = pd.read_csv(r'../data/WA_Fn-UseC_-HR-Employee-Attrition.csv')

# Drop useless cols
drop_cols = ['EmployeeCount', 'EmployeeNumber', 'Over18', 'StandardHours']
df = df.drop(columns=drop_cols)
```

```python

# Encode target
df['Attrition_encoded'] = LabelEncoder().fit_transform(df['Attrition'])
df = df.drop(columns=['Attrition'])   # drop original

# Binary encoding
df['Gender_encoded'] = LabelEncoder().fit_transform(df['Gender'])
df['OverTime_encoded'] = LabelEncoder().fit_transform(df['OverTime'])
df = df.drop(columns=['Gender', 'OverTime'])  # drop originals
```

```python

# One-hot encoding for multi-class categoricals
cols = ['BusinessTravel', 'Department', 'EducationField', 'JobRole', 'MaritalStatus']
df = pd.get_dummies(df, columns=cols, dtype=int)
```

```python

# Save processed
df.to_csv(r"../data/processed/cleaned.csv", index=False)

df.info()
```

```python
# ==========================================
# 🔽 Reduce feature set for simpler modeling
# ==========================================

# Keep only important + manageable features
selected_features = [
    "Age",
    "DistanceFromHome",
    "MonthlyIncome",
    "TotalWorkingYears",
    "YearsAtCompany",
    "JobSatisfaction",
    "EnvironmentSatisfaction",
    "WorkLifeBalance",
    "OverTime_encoded",
    "Gender_encoded",
    "MaritalStatus_Divorced",
    "MaritalStatus_Married",
    "MaritalStatus_Single"
]

# Create reduced DataFrame (with target)
df_reduced = df[selected_features + ["Attrition_encoded"]]

# Save for next pipeline step
df_reduced.to_csv("../data/processed/reduced_features.csv", index=False)

print("✅ Reduced feature dataset saved with shape:", df_reduced.shape)
df_reduced.head()
```
