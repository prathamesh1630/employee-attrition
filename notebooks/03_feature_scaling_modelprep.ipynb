```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import json
import os
```

```python

# ==========================================================
# 🔽 Load the reduced feature dataset (not full 50 features)
# ==========================================================
df = pd.read_csv("../data/processed/reduced_features.csv")

# Separate target and features
X = df.drop(columns=["Attrition_encoded"])
y = df["Attrition_encoded"]
```

```python

# ==========================================================
# 🔽 Scale numeric features
# ==========================================================
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Rebuild DataFrame with same column names
X_scaled_df = pd.DataFrame(X_scaled, columns=X.columns)
```

```python

# ==========================================================
# 🔽 Train/test split
# ==========================================================
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled_df, y, test_size=0.2, random_state=42, stratify=y
)

print("✅ Data ready for modeling")
print("Train shape:", X_train.shape, "Test shape:", X_test.shape)
```

```python

# ==========================================================
# 🔽 Save feature manifest (for Streamlit consistency)
# ==========================================================
manifest = {"features": list(X.columns)}
os.makedirs("../models", exist_ok=True)

with open("../models/feature_manifest.json", "w") as f:
    json.dump(manifest, f, indent=4)

print("✅ Feature manifest saved:", manifest)
```
